```python
# Load state-level weather dataset
df = pd.read_csv(
    r"C:\Users\Asus\OneDrive\Desktop\Sem 10 September 2026\Final Year Project 2\dataset\weather_all_states.csv"
)

# Ensure date column is parsed correctly
df["date"] = pd.to_datetime(df["date"]).dt.tz_localize(None)

# Create a Year-Month string column for heatmap x-axis
df["year_month"] = df["date"].dt.to_period("M").astype(str)

# Set visual theme
sns.set_theme(style="whitegrid")

# Function to generate and save heatmaps
def plot_weather_heatmap(data, value_col, title, cmap, filename):
    # Pivot table: rows = states, columns = year_month, values = metric
    pivot_df = data.pivot(index="state", columns="year_month", values=value_col)

    plt.figure(figsize=(16, 8))
    sns.heatmap(pivot_df, cmap=cmap, cbar_kws={'label': title}, linewidths=0.5, linecolor='white')
    
    plt.title(f"Heatmap of Monthly {title}", fontsize=14, fontweight="bold", pad=15)
    plt.xlabel("Year-Month", fontsize=11, fontweight="bold")
    plt.ylabel("States", fontsize=11, fontweight="bold")
    
    # Show fewer ticks on x-axis to prevent overcrowding (e.g., every 6 months)
    ax = plt.gca()
    for index, label in enumerate(ax.get_xticklabels()):
        if index % 6 != 0:
            label.set_visible(False)
            
    plt.xticks(rotation=45, ha="right")
    plt.tight_layout()
    plt.savefig(filename, dpi=300, bbox_inches="tight")
    plt.show()

# 1. Temperature Heatmap
plot_weather_heatmap(df, "temperature_C", "Temperature (°C)", "YlOrRd", "weather_heatmap_temperature.png")

# 2. Humidity Heatmap
plot_weather_heatmap(df, "humidity_pct", "Humidity (%)", "Blues", "weather_heatmap_humidity.png")

# 3. Rainfall Heatmap
plot_weather_heatmap(df, "rainfall_mm_per_day", "Rainfall (mm/day)", "YlGnBu", "weather_heatmap_rainfall.png")

# 4. Wind Speed Heatmap
plot_weather_heatmap(df, "wind_speed_m_s", "Wind Speed (m/s)", "Purples", "weather_heatmap_wind_speed.png")
```
